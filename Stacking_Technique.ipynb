### **Import Libraries**
This cell imports necessary libraries like pandas for data manipulation, numpy for numerical operations, and seaborn/matplotlib for data visualization.

```python
import pandas as pd;
import numpy as np;
import seaborn as sns;
import matplotlib.pyplot as plt;
```

### **Load Dataset**
This cell loads the diabetes dataset from a CSV file into a pandas DataFrame and displays the first few rows.

```python
df=pd.read_csv("https://raw.githubusercontent.com/Kirisaki00/Ensemble_Techniques-/refs/heads/main/diabetes.csv");
df.head()
```

### **Separate Features and Target**
Here, the dataset is split into features (x) and the target variable (y), where 'Outcome' is the target.

```python
x=df.drop(columns=["Outcome"], axis=1);
y=df["Outcome"];
```

### **Split Data into Train, Validation, and Test Sets**
This cell divides the data into training, validation, and test sets for model development and evaluation.

```python
from sklearn.model_selection import train_test_split;
train, val_train, test, val_test=train_test_split(x,y,test_size=0.5, random_state=42)
x_train, x_test, y_train, y_test=train_test_split(train, test, test_size=0.2, random_state=42)
```

### **Train and Evaluate K-Nearest Neighbors (KNN)**
This cell initializes, trains, and evaluates a K-Nearest Neighbors classifier on the training and test data.

```python
from sklearn.neighbors import KNeighborsClassifier;
knn=KNeighborsClassifier();
knn.fit(x_train, y_train);
knn.score(x_test, y_test)
```

### **Train and Evaluate Support Vector Machine (SVM)**
This cell initializes, trains, and evaluates a Support Vector Machine classifier.

```python
from sklearn.svm import SVC;
svm=SVC();
svm.fit(x_train,y_train);
svm.score(x_test, y_test)
```

### **Make Predictions on Validation Set**
This cell generates predictions from both KNN and SVM models on the validation training set.

```python
pred_val1=knn.predict(val_train);
pred_val2=svm.predict(val_train);
```

### **Display KNN Validation Predictions**
This cell displays the predictions made by the K-Nearest Neighbors model on the validation set.

```python
pred_val1
```

### **Display SVM Validation Predictions**
This cell displays the predictions made by the Support Vector Machine model on the validation set.

```python
pred_val2
```

### **Stack Validation Predictions**
This cell combines the predictions from KNN and SVM models on the validation set into a single array.

```python
pred_val=np.column_stack((pred_val1,pred_val2));
pred_val
```

### **Make and Stack Test Predictions**
This cell generates predictions from both models on the test set and stacks them.

```python
pred_test1=knn.predict(x_test);
pred_test2=svm.predict(x_test);
pred_test=np.column_stack((pred_test1,pred_test2));
pred_test
```

### **Train and Evaluate Random Forest (Meta-Learner)**
This cell trains a Random Forest Classifier using the stacked validation predictions as input and evaluates it on the stacked test predictions.

```python
from sklearn.ensemble import RandomForestClassifier;
rf=RandomForestClassifier();
rf.fit(pred_val, val_test);
rf.score(pred_test, y_test)
```

### **Setup GridSearchCV for Hyperparameter Tuning**
This cell defines the hyperparameter grid for tuning the Random Forest Classifier and initializes GridSearchCV.

```python
from sklearn.model_selection import GridSearchCV;
grid_parameters={
    "n_estimators":[90,100,115],
    "criterion":["gini", "entropy"],
    "min_samples_leaf":[1,2,3,4,5],
    "max_features":[4,5,6,7,8]
};
grid_search=GridSearchCV(estimator=rf, param_grid=grid_parameters, cv=5);
```

### **Fit GridSearchCV and Get Best Parameters**
This cell executes the grid search to find the best hyperparameters for the Random Forest model based on the validation data.

```python
grid_search.fit(pred_val, val_test);
grid_search.best_params_
```

### **Train and Evaluate Random Forest with Tuned Hyperparameters**
This cell trains a new Random Forest Classifier using the best hyperparameters found by GridSearchCV and evaluates its performance on the test set.

```python
from sklearn.ensemble import RandomForestClassifier;
rf=RandomForestClassifier(criterion='gini',max_features= 4,min_samples_leaf=1,n_estimators=90);
rf.fit(pred_val, val_test);
rf.score(pred_test, y_test)
```
